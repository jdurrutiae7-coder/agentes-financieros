# Análisis cuantitativo en VS Code

Selecciona el kernel `.venv`. Primero prepara el entorno siguiendo el README. Este notebook usa el mismo motor que la terminal. La única configuración inicial está en la primera celda. DEMO viene activado para probar sin descargar datos.


```python
TICKER = "AAPL"
BENCHMARK = "SPY"
DEMO = True
```

```python
from pathlib import Path
import subprocess
import sys

RAIZ = Path.cwd().resolve()
if not (RAIZ / "analisis.py").exists():
    RAIZ = RAIZ.parent
if not (RAIZ / "analisis.py").exists():
    raise RuntimeError("Abre el notebook dentro de la carpeta agentes-financieros.")
subprocess.run([sys.executable, str(RAIZ / "analisis.py"), "crear", "--ticker", TICKER, "--benchmark", BENCHMARK], cwd=RAIZ, check=True)
print("Revisa empresas/" + TICKER + "/config.json antes de correr con datos reales.")
```

```python
comando = [sys.executable, str(RAIZ / "analisis.py"), "ejecutar", "--empresa", TICKER, "--benchmark", BENCHMARK]
if DEMO:
    comando.append("--demo")
proceso = subprocess.run(comando, cwd=RAIZ, check=True, capture_output=True, text=True, encoding="utf-8")
print(proceso.stdout)
if proceso.stderr:
    print(proceso.stderr)
```

```python
import json
from IPython.display import Markdown, Image, display

carpeta = Path(proceso.stdout.strip().splitlines()[-1])
display(Markdown((carpeta / "resumen.md").read_text(encoding="utf-8")))
for grafico in sorted(carpeta.glob("*.png")):
    display(Image(filename=str(grafico)))
```
